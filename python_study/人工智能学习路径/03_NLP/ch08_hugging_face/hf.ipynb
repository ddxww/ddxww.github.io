## 2.加载和使用Tokenizer

### 2.1加载

```python
from transformers import AutoTokenizer

# 加载分词器
tokenizer = AutoTokenizer.from_pretrained("google-bert/bert-base-chinese")
```

### 2.2使用

```python
# 1.分词（tokenize）
tokens = tokenizer.tokenize("我爱自然语言处理")
tokens
```

```python
# 2.token->id
ids = tokenizer.convert_tokens_to_ids(tokens)
ids
```

```python
# 3.id->token
tokens = tokenizer.convert_ids_to_tokens(ids)
tokens
```

```python
# 4.encode
ids = tokenizer.encode("我爱自然语言处理", padding='max_length', max_length=20)
ids
```

```python
text = tokenizer.decode(ids, skip_special_tokens=True)
text.replace(" ", "")
```

```python
# 5.__call__
inputs = tokenizer("我爱自然语言处理")
inputs
```

### 2.3分词模型配合使用

```python
from transformers import AutoModel, AutoTokenizer
import torch

texts = ["我爱自然语言处理", "我爱人工智能", "我们一起学习"]

model = AutoModel.from_pretrained("google-bert/bert-base-chinese")
tokenizer = AutoTokenizer.from_pretrained("google-bert/bert-base-chinese")

inputs = tokenizer(texts, padding=True, return_tensors='pt')

with torch.no_grad():
    output = model(**inputs)

print(output.last_hidden_state.shape)
print(output.pooler_output.shape)
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

texts = ["我爱自然语言处理", "我爱人工智能", "我们一起学习"]

model = AutoModelForSequenceClassification.from_pretrained("google-bert/bert-base-chinese")
tokenizer = AutoTokenizer.from_pretrained("google-bert/bert-base-chinese")

inputs = tokenizer(texts, padding=True, return_tensors='pt')

with torch.no_grad():
    output = model(**inputs)

print(output)
print(output.logits.shape)
```

### 3.3预处理数据集

```python
# 1.删除列
dataset = dataset.remove_columns(['cat'])
```

```python
dataset.features
```

```python
# 2.过滤行
dataset = dataset.filter(lambda x: x['review'] is not None)
```

```python
# 3.划分数据集
dataset_dict = dataset.train_test_split(test_size=0.2)
dataset_dict
```

```python
dataset_dict['train']
```

```python
# 4.map操作（逐条）
train_dataset = dataset_dict['train']
tokenizer = AutoTokenizer.from_pretrained("google-bert/bert-base-chinese")


def encode(example):
    # example:{label:1.review::'******'}
    inputs = tokenizer(example['review'], padding='max_length', max_length=128, truncation=True)
    inputs['labels'] = example['label']
    return inputs

train_dataset=train_dataset.map(encode, batched=False,remove_columns=['label','review'])
train_dataset.features
```
