```python
import torch
```

```python
tensor1=torch.randint(1,10,(3,2,4)).float()
print(tensor1)
```

```python
#sum求和
print(tensor1.sum())
```

```python
print(tensor1.sum(dim=2))
```

```python
#mean求均值
print(tensor1.mean(dim=2))
```

```python
#求标准方差
print(tensor1.std())
```

```python
# max,min求最大最小值
print(tensor1.min( dim=2))
```

```python
print(tensor1.argmin())
```

```python
print(tensor1.argmax())
```

```python
print(tensor1.unique())
```

```python
print(tensor1.sort(dim=2))
```
