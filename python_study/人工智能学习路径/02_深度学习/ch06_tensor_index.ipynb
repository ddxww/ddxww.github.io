```python
import torch
```

```python
tensor1=torch.randint(1,10,(3,5,4))
print(tensor1)
```

```python
# 1.简单索引
print(tensor1[2,1,3])
print(tensor1[2][1][3])
```

```python
print(tensor1[2,3])
print(tensor1[2][3])
```

```python
print(tensor1[1])
```

```python
print(tensor1[:,3])
```

```python
# 2.范围索引
print(tensor1[1:])
```

```python
print(tensor1[-1:,1:4:2,0:3:2])
```

```python
print(tensor1[-1:,-4:-1:2,0:3:2])
```

```python
#列表索引
print(tensor1[[0,1],[1,2]]) #第0矩阵的第1行，第1矩阵的第2行
```

```python
print(tensor1[[0,1],[3,2],[2,3]])
```

```python
print(tensor1[[[0],[1]],[1,2]])
```

```python
#布尔索引
#选取符合条件的行，首元素大于5
mask=tensor1[:,:,0]>5
print(mask)
```

```python
print(tensor1[mask])
```

```python
#选取符合条件的行，第二个元素大于5
mask=tensor1[:,1,:]>5
print(mask)
```

```python
# print(tensor1[mask])
tensor2=tensor1.mT
print(tensor2)
```

```python
#将mask运用到tensor2上，形状匹配前两个维度
print(tensor2[mask])
```

```python
print(tensor2[mask].mT)
print(tensor2[mask].T)
```

```python
#选取符合条件的矩阵，元素(1,2)>5
mask=tensor1[:,1,2]>5
print(mask)
print(tensor1[mask])
```

```python
mask=tensor1>5
print(mask)
```

```python
print(tensor1[mask])
```
